# Sales Data Calculator 📊💰

## Task 4 - Level 1 - Day 4

This project analyzes a collection of sales values using Python.

The program calculates:

- Total sales
- Average sales
- Highest sale
- Lowest sale

```python
# Store sales values in a Python list

sales = [
    1200,
    1500,
    980,
    1750,
    2200,
    1450,
    1900,
    1250,
    1600,
    2500
]

print("Sales Values:", sales)
```

## 1. Total Sales

The `sum()` function is used to calculate the total sales value.

```python
# Calculate total sales

total_sales = sum(sales)

print("Total Sales:", total_sales)
```

## 2. Average Sales

The average sales value is calculated by dividing total sales by the number of sales.

### Formula

Average Sales = Total Sales / Number of Sales

```python
# Calculate average sales

average_sales = total_sales / len(sales)

print("Average Sales:", round(average_sales, 2))
```

## 3. Highest Sale

The `max()` function is used to find the highest sale value.

```python
# Find the highest sale

highest_sale = max(sales)

print("Highest Sale:", highest_sale)
```

## 4. Lowest Sale

The `min()` function is used to find the lowest sale value.

```python
# Find the lowest sale

lowest_sale = min(sales)

print("Lowest Sale:", lowest_sale)
```

## 5. Final Sales Analysis

The following code displays all calculated sales statistics together.

```python
# Display final sales analysis

print("======================================")
print("          SALES DATA CALCULATOR")
print("======================================")

print("Number of Sales:", len(sales))
print("Total Sales:", total_sales)
print("Average Sales:", round(average_sales, 2))
print("Highest Sale:", highest_sale)
print("Lowest Sale:", lowest_sale)
```
